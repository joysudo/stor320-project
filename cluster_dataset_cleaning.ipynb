```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("spotify_most_streamed_songs.csv")
print(df.info())
print(df.isna().sum())
print(df.duplicated().sum())
print(df.head())
print(df.columns.to_list())

```

```python
df["streams"] = pd.to_numeric(df["streams"], errors="coerce")
df = df.dropna(subset=["streams"]).copy()


df = df.drop_duplicates(
    subset=["track_name", "artist(s)_name"]
).copy()


features = ["bpm", "danceability_%", "valence_%", "energy_%", "acousticness_%", "instrumentalness_%", "liveness_%", "speechiness_%"]


df = df.dropna(subset=features).copy()
df = df.reset_index(drop=True)


X = df[features].copy()


scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


df_scaled = pd.DataFrame(
    X_scaled,
    columns=features,
    index=df.index
)


df.to_csv("spotify_cluster_cleaned.csv", index=False)
df_scaled.to_csv("spotify_cluster_scaled.csv", index=False)

print(df_scaled.head())
```
