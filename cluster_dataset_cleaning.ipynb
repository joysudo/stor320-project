```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("featuresdf.csv")
print(df.info())
print(df.isna().sum())
print(df.duplicated().sum())
print(df.head())
```

```python
features = ["danceability", "energy", "key", "loudness", "speechiness", "acousticness", "instrumentalness", "liveness", "valence", "tempo"]

X = df[features].copy()

#scaling because different numeric values were at diff ranges
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


df_scaled = pd.DataFrame(
    X_scaled,
    columns=features,
    index=df.index
)

df.to_csv("spotify_cleaned.csv", index=False)
df_scaled.to_csv("spotify_scaled.csv", index=False)

print(df_scaled.head())
```
