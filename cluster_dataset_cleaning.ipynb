```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("dataset.csv")
print(df.info())
print(df.isna().sum())
print(df.duplicated().sum())
print(df.head())
print(df.columns.to_list())

```

```python
df = df.drop_duplicates(subset=["track_id"]).copy()


features = ["acousticness", "danceability", "duration_ms", "energy", "instrumentalness", "liveness", "loudness", "speechiness", "tempo", "valence"]


df = df.dropna(subset=features + ["popularity"]).copy()


df = df.reset_index(drop=True)


X = df[features].copy()


scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


df_scaled = pd.DataFrame(
    X_scaled,
    columns=features,
    index=df.index
)


df.to_csv("spotify_cluster_cleaned.csv", index=False)
df_scaled.to_csv("spotify_cluster_scaled.csv", index=False)


print("Cleaned dataset:", df.shape)
print("Scaled dataset:", df_scaled.shape)
print(df_scaled.head())
```
